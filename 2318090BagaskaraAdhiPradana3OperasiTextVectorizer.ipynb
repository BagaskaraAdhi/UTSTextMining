# **1. Mount Google Drive & Load Dataset**

```python
from google.colab import drive
import pandas as pd
import numpy as np

# Menyambungkan ke Google Drive
drive.mount('/content/drive')

# Mengambil dataset dari path UTS kamu
path = '/content/drive/MyDrive/Text Mining/UTS/Dataset/GojekAppReviewV4.0.0-V4.9.3_Cleaned.csv'
df = pd.read_csv(path)

# MENAMPILKAN 10 DATA PERTAMA DARI DATASET ASLI
print("--- 10 Baris Pertama Dataset Asli ---")
display(df.head(10))

# Preprocessing: Menghapus data kosong dan sampling 1000 data
df = df.dropna(subset=['content'])
documents = df['content'].head(1000).astype(str).tolist()

print(f"\nBerhasil memuat {len(documents)} dokumen untuk diproses.")
print("=" * 60)

output_path = '/content/drive/MyDrive/Text Mining/UTS/Dataset/'
```

# **2. Bag Of Words (BOW)**

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer_bow = CountVectorizer()
vectors_bow = vectorizer_bow.fit_transform(documents)
feature_names_bow = vectorizer_bow.get_feature_names_out()

# Tampilkan 10 data/kolom pertama di Colab
print("--- Hasil Bag Of Words (10 Kolom Pertama) ---")
print("Fitur:", feature_names_bow[:10])
print(vectors_bow.toarray()[:10, :10]) # 10 baris x 10 kolom

# Simpan ke Excel
df_bow = pd.DataFrame(vectors_bow.toarray(), columns=feature_names_bow)
df_bow.to_excel('Hasil_BoW_Gojek.xlsx', index=False)
print("\n> File 'Hasil_BoW_Gojek.xlsx' berhasil dibuat.")
df_bow.to_excel(output_path + 'Hasil_BoW_Gojek.xlsx', index=False)
print(f"\n> File 'Hasil_BoW_Gojek.xlsx' tersimpan di: {output_path}")
```

# **3. TF-ID**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer_tfidf = TfidfVectorizer()
vectors_tfidf = vectorizer_tfidf.fit_transform(documents)
feature_names_tfidf = vectorizer_tfidf.get_feature_names_out()

# Tampilkan 10 data/kolom pertama di Colab
print("--- Hasil TF-IDF (10 Kolom Pertama) ---")
print("Fitur:", feature_names_tfidf[:10])
print(vectors_tfidf.toarray()[:10, :10]) # 10 baris x 10 kolom

# Simpan ke Excel
df_tfidf = pd.DataFrame(vectors_tfidf.toarray(), columns=feature_names_tfidf)
df_tfidf.to_excel('Hasil_TFIDF_Gojek.xlsx', index=False)
print("\n> File 'Hasil_TFIDF_Gojek.xlsx' berhasil dibuat.")
df_tfidf.to_excel(output_path + 'Hasil_TFIDF_Gojek.xlsx', index=False)
print(f"> File 'Hasil_TFIDF_Gojek.xlsx' tersimpan di: {output_path}")
```

# **4. Word Embedding (Word2Vec)**

```python
from gensim.models import Word2Vec

# Tokenisasi (Word2Vec butuh list kata)
sentences = [text.lower().split() for text in documents]

# Training model Word2Vec (100 dimensi sesuai modul)
model_w2v = Word2Vec(sentences, vector_size=100, window=5, min_count=1, workers=4)

# Tampilkan 10 dimensi pertama dari kata 'gojek'
print("--- Hasil Word Embedding (10 Dimensi Pertama kata 'gojek') ---")
if 'gojek' in model_w2v.wv:
    print(model_w2v.wv['gojek'][:10])

# Simpan hasil Word Vectors ke Excel (Kata -> Vektor 100 Dimensi)
words = list(model_w2v.wv.index_to_key)
word_vectors = [model_w2v.wv[word] for word in words]
df_w2v = pd.DataFrame(word_vectors, index=words)
df_w2v.to_excel('Hasil_Word2Vec_Gojek.xlsx')
print("\n> File 'Hasil_Word2Vec_Gojek.xlsx' berhasil dibuat.")
df_w2v.to_excel(output_path + 'Hasil_Word2Vec_Gojek.xlsx')
print(f"> File 'Hasil_Word2Vec_Gojek.xlsx' tersimpan di: {output_path}")
```
